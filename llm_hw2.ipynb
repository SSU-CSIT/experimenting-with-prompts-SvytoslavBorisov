## Experimenting with Prompts

Large language models (LLMs) are pre-trained on vast amounts of data to generate novel content, including text, images, and video. To generate the desired response, you should feed the input to a model as a prompt. It is important to provide a context-based, clear, and precise prompt to guide the model and get the desired output. The process of designing effective prompts to generate better and desired responses is called prompt engineering.

In this lab, you will learn the basics of prompt engineering and experiment with simple prompts to control how a model interprets and generates desired and coherent responses.

### Learning Objectives:
* Design a prompt structure to write concise prompts based on requirements.
* Write unambiguous and context-based prompts to generate the desired output.
* Write prompts in different formats, such as questions, statements, and instructions.
* Let's get started prompting and experimenting with foundation models.

## Exercise 3: Prompts in different formats

Prompts can be presented in different formats, such as questions, statements, or instructions. Let's experiment with how prompts in different formats can yield different results.

First, let's write a prompt in question format. Enter the prompt

```python
What are the benefits of water reservoirs in a detailed paragraph?
```

### Prompt 1 (Question format):


```python
What are the benefits of water reservoirs in a detailed paragraph?
```

### Response 1:


```python
Water reservoirs provide numerous benefits. They store water during rainy seasons to ensure a steady supply during dry periods, preventing water shortages. Reservoirs also help in flood control by regulating river flow, support irrigation for agriculture, and serve as sources of drinking water. Additionally, they contribute to hydropower generation and create habitats for aquatic life.
```

Next, write the prompt in statement format. 

```python
Discuss the benefits of utilizing water reservoirs.
```

### Prompt 2 (Statement format):

```python
Discuss the benefits of utilizing water reservoirs.
```

### Response 2:


```python
Utilizing water reservoirs offers many advantages, including water storage for domestic and industrial use, agricultural irrigation, flood prevention, and electricity generation through hydropower. Reservoirs also promote environmental stability and recreational opportunities like fishing and boating.
```

Further, write the prompt in an instruction format.

List the top five benefits of water reservoirs.

### Prompt 3 (Instruction format):


```python
List the top five benefits of water reservoirs.
```

### Response 3:


```python
1. Water storage for use during dry seasons  
2. Flood control and river flow regulation  
3. Irrigation for agriculture  
4. Hydropower generation  
5. Recreational and ecological benefits
```

You can review and analyze the response generated by each prompt to identify which prompt format works best to accomplish the desired task. For example, the instruction format would work well for a summarization task. For text completion, the statement format would be suitable. When you want to extract a specific entity from a statement, the question format would be most suitable.

### Сравнение:

- Каждый формат промпта влияет на структуру и акценты в ответе модели:

- Формат вопроса дал подробное объяснение в виде абзаца.

- Формат утверждения предоставил сбалансированный, умеренно детализированный ответ, подходящий для эссе или отчётов.

- Формат инструкции выдал краткий и чёткий список — идеально подходящий для задач на суммирование или презентацию.

Использование разных форматов промптов помогает направить модель к нужному уровню детализации и структуре ответа, соответствующей цели задания.
